## FIR filter

```python
# Libraries
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Reading the frames from the video

vidcap = cv2.VideoCapture('./v_Biking_g14_c04.avi')
success, image = vidcap.read()

frames_arr = []
while success:
    success, image = vidcap.read()
    
    if success == False:
        break
        
    #gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    frames_arr.append(image)
```

```python
len(frames_arr)
```

### Temporal Smoothing

```python
#tmp_smooth = []

# Defining frames per second and size of the video
fps = vidcap.get(cv2.CAP_PROP_FPS)
frame_width = int(vidcap.get(3)) 
frame_height = int(vidcap.get(4)) 
size = (frame_width, frame_height)

# Defining the video file
result = cv2.VideoWriter('/Users/sofia.lucia/Documents/VR/smooth.mp4', cv2.VideoWriter_fourcc('m','p','4','v'), fps, size)

# Saving the result 
for n in range(2,len(frames_arr)):
    # Applying the temporal smoothing
    img_smooth = np.mean(frames_arr[n - 2:n+1], axis=0)
    # Normalization
    img_smooth *= 255.0 / np.max(img_smooth)
    #tmp_smooth.append(img_smooth)
    # 
    result.write(img_smooth.astype('uint8'))
    
result.release()
```

### Temporal differentiating 

```python
def sobel_magnitude(image):
    """
    
    """
    sobel_x = cv2.Sobel(image, ddepth=cv2.CV_32F, dx=1, dy=0, ksize=3)  # horizontal gradient
    sobel_y = cv2.Sobel(image, ddepth=cv2.CV_32F, dx=0, dy=1, ksize=3)  # vertical gradient
    magnitude = np.sqrt(sobel_x**2 + sobel_y**2)
    magnitude *= 255.0 / np.max(magnitude)  # normalization
    
    return magnitude

    
def temporal_diff(frame_1, frame_2):
    magnitude_fr_1 = sobel_magnitude(frame_1)
    magnitude_fr_2 = sobel_magnitude(frame_2)
    output = magnitude_fr_1 - magnitude_fr_2
    
    return output
```

```python
#fir_frames_diff = []
# Defining the video file
result = cv2.VideoWriter('/Users/sofia.lucia/Documents/VR/output_fir_diff.mp4', cv2.VideoWriter_fourcc('m','p','4','v'), fps, size)

for n_frame in range(2, len(frames_arr)):
    # Temporal difference
    diff_sobel = temporal_diff(frames_arr[n_frame], frames_arr[n_frame - 2])
    # Normalization
    diff_sobel *= 255.0 / np.max(diff_sobel)
    # Writing the frame
    result.write(diff_sobel.astype('uint8'))
    #fir_frames_diff.append(diff_sobel)
        
result.release()
```
